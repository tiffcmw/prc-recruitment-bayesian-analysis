```python
from prc1_state import State, SortedSetAndComplement
from prc1 import Prc1
from lattice import Lattice
from run_gillespie import run_gillespie_prc1
import numpy as np
import matplotlib.pyplot as plt
from gillespie import run_gillespie
from copy import deepcopy
from gillespie import *
from run_gillespie import *
import runpy, os
from sortedcontainers import SortedSet
import matplotlib.pyplot as plt
```

```python
lattice = Lattice(microtubule_length=30,
                  microtubule_width=30,
                  microtubule_spacing=5,
                  microtubule_separation=10,
                  microtubule_offset=0)
```

```python
n = 0

rows = []
cols = []

for idx in range(lattice.num_sites):
    r, c = lattice.to_coord(idx)
    rows.append(r)
    cols.append(c)

row, col = lattice.to_coord(n)
nbrs = lattice.get_neighbors(n)
nbr_row = []
nbr_col = []
for nbr in nbrs:
    r, c = lattice.to_coord(nbr)
    nbr_row.append(r)
    nbr_col.append(c)

# Plot
plt.figure(figsize=(10, 4))
plt.scatter(cols, rows, s=20, c='blue')
plt.scatter(col, row, color='red', s=50, zorder=5)
plt.scatter(nbr_col, nbr_row, color='orange', s=50, zorder=5)
plt.title("Microtubule 2D Lattice Sites")
plt.xlabel("Column index")
plt.ylabel("Row index")
plt.show()
```
